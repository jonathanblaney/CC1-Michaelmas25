### Calculate softmax



```python
import math
```

This notebook takes a list of numbers and calculates probabilities for each, such that the overall total is 1.0


We first compute the total for each *n* in the list as
\begin{equation}
 \epsilon^{n}
\end{equation}
For example, for the numbers 2, 7, 1
\begin{equation}
\epsilon^{2} + \epsilon^{7} + \epsilon^{1}
\end{equation}

```python
e = math.e
print(e)
```

```python
e**3
```

```python
probabilites_list = [5, 4, -1]
softmax_total = []
final_softmax = []
```

```python
for probability in probabilites_list:
  print(f"{probability} becomes {e**probability}")
  softmax_total.append(e**probability)
```

```python
overall_total = sum(softmax_total)
print(overall_total)
```

```python
for probability in softmax_total:
  print(f"{probability} becomes {probability/overall_total:.2}")
  final_softmax.append(probability/overall_total)
```

```python
for one, two in zip(probabilites_list, final_softmax):
  print(f"{one} => {two}")
```
